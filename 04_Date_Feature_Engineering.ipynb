## converting date of submission nd dos 

```python
# Convert date columns to datetime
claims["Service_Date"] = pd.to_datetime(claims["Service_Date"])
claims["Claim_Submission_Date"] = pd.to_datetime(claims["Claim_Submission_Date"])

# Extract month
claims["Service_Month"] = claims["Service_Date"].dt.month
claims["Submission_Month"] = claims["Claim_Submission_Date"].dt.month

# Extract day of week
claims["Service_DayOfWeek"] = claims["Service_Date"].dt.dayofweek
claims["Submission_DayOfWeek"] = claims["Claim_Submission_Date"].dt.dayofweek

# Calculate claim submission delay
claims["Claim_Submission_Delay"] = (
    claims["Claim_Submission_Date"] - claims["Service_Date"]
).dt.days
```
