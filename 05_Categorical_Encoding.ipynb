## Label Encoding

```python
categorical_features = [
    "Claim_Form_Type",
    "Provider_Specialty",
    "Payer",
    "Diagnosis_Code",
    "Modifier_Code"
]
categorical_features
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoders = {}

for column in categorical_features:
    le = LabelEncoder()
    claims[column] = le.fit_transform(claims[column])

    label_encoders[column] = le
```

```python
claims[categorical_features].head()
```

```python
import joblib

joblib.dump(label_encoders, "label_encoders.pkl")

print("Label Encoders Saved Successfully!")
```
