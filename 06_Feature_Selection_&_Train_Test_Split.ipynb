```python
import pandas as pd
import joblib

claims = pd.read_csv("claims_preprocessed_v3.csv")

claims.head()
```

## Create the Feature Matrix (X)

```python
drop_columns = [
    "Claim_Status",
    "CARC_Code",
    "Denial_Reason",
    "Procedure_Desc",
    "Place_of_Service_Desc",
    "Service_Date",
    "Claim_Submission_Date"
]

X = claims.drop(columns=drop_columns)
```

```python
X
```

## Create the Target (y)

```python
y = claims["Claim_Status"]
```

## Claim_Status is still text (Approved / Denied), we'll encode

```python
from sklearn.preprocessing import LabelEncoder

target_encoder = LabelEncoder()

y = target_encoder.fit_transform(y)
```

```python
print(dict(zip(target_encoder.classes_,
               target_encoder.transform(target_encoder.classes_))))
```

```python
joblib.dump(target_encoder, "target_encoder.pkl")
```
