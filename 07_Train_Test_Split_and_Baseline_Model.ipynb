```python
import pandas as pd

claims = pd.read_csv("claims_preprocessed_v3.csv")
```

```python
import joblib

label_encoders = joblib.load("label_encoders.pkl")
target_encoder = joblib.load("target_encoder.pkl")
```

```python
drop_columns = [
    "Claim_Status",
    "CARC_Code",
    "Denial_Reason",
    "Procedure_Desc",
    "Place_of_Service_Desc",
    "Service_Date",
    "Claim_Submission_Date"
]

X = claims.drop(columns=drop_columns)
```

```python
y = claims["Claim_Status"]
y = target_encoder.transform(y)
```

```python
print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nFirst 5 Features")

X.head()
```

## Check Class Distribution

```python
print(claims["Claim_Status"].value_counts())

print("\nPercentage")

print(claims["Claim_Status"].value_counts(normalize=True) * 100)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Set")

print(X_train.shape)
print(y_train.shape)

print()

print("Testing Set")

print(X_test.shape)
print(y_test.shape)
```

```python
import joblib

joblib.dump(X_train, "X_train.pkl")
joblib.dump(X_test, "X_test.pkl")
joblib.dump(y_train, "y_train.pkl")
joblib.dump(y_test, "y_test.pkl")

print("Train-Test Split Saved Successfully!")
```

```python
X_train.to_csv("X_train.csv", index=False)

print("X_train.csv saved successfully!")
X_test.to_csv("X_test.csv", index=False)

print("X_test.csv saved successfully!")
```

```python
import pandas as pd

pd.DataFrame(
    y_train,
    columns=["Claim_Status"]
).to_csv("y_train.csv", index=False)

print("y_train.csv saved successfully!")
```

```python
pd.DataFrame(
    y_test,
    columns=["Claim_Status"]
).to_csv("y_test.csv", index=False)

print("y_test.csv saved successfully!")
```
